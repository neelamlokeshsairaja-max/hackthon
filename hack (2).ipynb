```python
#data and time columns
date_columns = [
    'flight_date',
    'scheduled_departure',
    'actual_departure',
    'scheduled_arrival',
    'actual_arrival'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

df[date_columns].dtypes
```

data quality checks

```python
#missing values
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    'Column': missing.index,
    'Missing_Values': missing.values,
    'Missing_Percentage': (missing.values / len(df) * 100).round(2)
})

missing_df[missing_df['Missing_Values'] > 0]
```

```python
#check dupliacte records
print("Total duplicates:", df.duplicated().sum())

print("Duplicate Flight IDs:",
      df['flight_id'].duplicated().sum())
```

```python
#convert data columns
date_columns = [
    'flight_date',
    'scheduled_departure',
    'actual_departure',
    'scheduled_arrival',
    'actual_arrival'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
#crate caulated col
#toatl revinue
df['total_revenue'] = (
    df['ticket_revenue'] +
    df['ancillary_revenue']
)
```

```python
#total cost
df['total_cost'] = (
    df['fuel_cost'] +
    df['crew_cost'] +
    df['airport_charges'] +
    df['maintenance_cost']
)
```

```python
#profit
df['profit'] = (
    df['total_revenue'] -
    df['total_cost']
)
```

```python
#passemger loading
df['load_factor'] = (
    df['seats_sold'] /
    df['total_seats']
) * 100
```

```python
#dealy depature
df['departure_delay'] = (
    df['actual_departure'] -
    df['scheduled_departure']
).dt.total_seconds() / 60
```

```python
#verify new col
df[[
    'flight_id',
    'total_revenue',
    'total_cost',
    'profit',
    'load_factor',
    'departure_delay'
]].head(10)
```

```python
#check the skylink airway perfromance
print("Total Revenue:", df['total_revenue'].sum())

print("Total Cost:", df['total_cost'].sum())

print("Total Profit:", df['profit'].sum())

print("Average Load Factor:",
      df['load_factor'].mean())

print("Average Departure Delay:",
      df['departure_delay'].mean())
```

```python
#visulisation library
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
```

```python
#flight stat analysis
status = df['flight_status'].value_counts()

plt.figure(figsize=(7, 5))
plt.pie(
    status,
    labels=status.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Flight Status Distribution')
plt.show()
```

```python
#flight dealy reasons
delay_data = df[df['delay_reason'] != 'None']

plt.figure(figsize=(9, 5))
sns.countplot(
    data=delay_data,
    x='delay_reason',
    order=delay_data['delay_reason'].value_counts().index
)
plt.title('Flight Delay Reasons')
plt.xlabel('Delay Reason')
plt.ylabel('Number of Flights')
plt.xticks(rotation=30)
plt.show()
```

```python
#delay by route
route_delay = (
    df[df['flight_status'] == 'Operated']
    .groupby('route')['departure_delay']
    .mean()
    .sort_values(ascending=False)
)

route_delay.plot(kind='bar', figsize=(12, 5))

plt.title('Average Departure Delay by Route')
plt.xlabel('Route')
plt.ylabel('Average Delay (Minutes)')
plt.xticks(rotation=45)
plt.show()
```

```python
#passenger loading
route_load = (
    df[df['flight_status'] == 'Operated']
    .groupby('route')['load_factor']
    .mean()
    .sort_values()
)

route_load.plot(kind='bar', figsize=(12, 5))

plt.axhline(60, linestyle='--',
            label='60% threshold')

plt.title('Average Passenger Load Factor by Route')
plt.xlabel('Route')
plt.ylabel('Load Factor (%)')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
#route profit analaysis
route_profit = (
    df.groupby('route')['profit']
    .sum()
    .sort_values(ascending=False)
)

route_profit.plot(kind='bar', figsize=(12, 5))

plt.title('Total Profit by Route')
plt.xlabel('Route')
plt.ylabel('Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
#montly analysis
df['month'] = df['flight_date'].dt.month

monthly_revenue = (
    df.groupby('month')['total_revenue']
    .sum()
)

monthly_revenue.plot(
    kind='line',
    marker='o',
    figsize=(10, 5)
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(range(1, 13))
plt.show()
```

```python
#morning and evining delay
import numpy as np
df['flight_period'] = np.where(
    df['scheduled_departure'].dt.hour < 12,
    'Morning',
    'Evening'
)

period_delay = (
    df[df['flight_status'] == 'Operated']
    .groupby('flight_period')['departure_delay']
    .mean()
)

period_delay.plot(
    kind='bar',
    figsize=(7, 5)
)

plt.title('Morning vs Evening Average Delay')
plt.xlabel('Flight Period')
plt.ylabel('Average Delay (Minutes)')
plt.xticks(rotation=0)
plt.show()
```

```python
#delay distribution
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df[df['flight_status'] == 'Operated'],
    x='departure_delay',
    bins=30,
    kde=True
)

plt.title('Distribution of Flight Delays')
plt.xlabel('Departure Delay (Minutes)')
plt.ylabel('Flight Count')
plt.show()
```

```python
#2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"/content/DA10_Cleaned_Aviation_Dataset.csv")

print(df.head())
```

```python
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
df['flight_date'] = pd.to_datetime(df['flight_date'], errors='coerce')

df = df.dropna(subset=['flight_date', 'route'])

df['month'] = df['flight_date'].dt.month_name()

df['month_number'] = df['flight_date'].dt.month
```

```python
monthly_analysis = df.groupby(
    ['month_number', 'month', 'route']
).agg(
    total_flights=('flight_id', 'count'),
    total_passengers=('seats_sold', 'sum'),
    total_revenue=('total_revenue', 'sum')
).reset_index()

monthly_analysis = monthly_analysis.sort_values(
    ['month_number', 'route']
)

print(monthly_analysis)
```

```python
monthly_passengers = df.groupby('month')['seats_sold'].sum()

monthly_passengers.plot(
    kind='bar',
    figsize=(10,5),
    title='Monthly Passenger Analysis'
)

plt.xlabel('Month')
plt.ylabel('Total Passengers')
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_revenue = df.groupby('month')['total_revenue'].sum()

monthly_revenue.plot(
    kind='line',
    marker='o',
    figsize=(10,5),
    title='Monthly Revenue Trend'
)

plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
route_revenue = df.groupby('route')['total_revenue'].sum().sort_values(ascending=False)

route_revenue.head(10).plot(
    kind='bar',
    figsize=(10,5),
    title='Top 10 Routes by Revenue'
)

plt.xlabel('Route')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_summary = df.groupby('month').agg(
    total_flights=('flight_id', 'count'),
    total_passengers=('seats_sold', 'sum'),
    total_revenue=('total_revenue', 'sum')
)

print(monthly_summary)
```

```python
correlation = df[
    ['seats_sold', 'total_revenue', 'total_cost', 'profit', 'load_factor']
].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Aviation Business Correlation Analysis')
plt.show()
```

```python
monthly_analysis.to_excel(
    "Aviation_Monthly_EDA.xlsx",
    index=False
)

print("EDA analysis completed successfully!")
```
